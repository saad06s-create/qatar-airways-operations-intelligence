```python
conn = sqlite3.connect("../sql/qatar_airways.db")
cursor = conn.cursor()

print("Database created successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Flights (

    flight_id INTEGER PRIMARY KEY,

    flight_date TEXT,

    airline TEXT,

    flight_number TEXT,

    origin TEXT,

    destination TEXT,

    scheduled_departure INTEGER,

    actual_departure INTEGER,

    departure_delay INTEGER,

    arrival_delay INTEGER,

    distance INTEGER,

    aircraft TEXT

)
""")

conn.commit()

print("Flights table created ✈️")
```
