```python
import numpy as np

# Create a proper date
flights["flight_date"] = (
    flights["year"].astype(str)
    + "-"
    + flights["month"].astype(str).str.zfill(2)
    + "-"
    + flights["day"].astype(str).str.zfill(2)
)

# On-time KPI
flights["is_delayed"] = flights["dep_delay"] > 15

# Route classification
flights["route_type"] = np.where(
    flights["distance"] >= 1500,
    "Long Haul",
    "Short Haul"
)

flights.head()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../sql/qatar_airways.db")

print("Connected successfully ✈️")
```

```python
query = """
SELECT
    COUNT(*) AS total_flights,
    SUM(CASE WHEN dep_delay <= 15 THEN 1 ELSE 0 END) AS on_time
FROM Flights
"""

kpi = pd.read_sql(query, conn)

total = int(kpi.loc[0, "total_flights"])
on_time = int(kpi.loc[0, "on_time"])

otp = (on_time / total) * 100

print(f"On-time Performance: {otp:.1f}%")
```

```python
query = """
SELECT
    origin,
    ROUND(AVG(dep_delay),2) AS avg_delay,
    COUNT(*) AS flights
FROM Flights
GROUP BY origin
HAVING flights > 1000
ORDER BY avg_delay DESC
"""

airport_delays = pd.read_sql(query, conn)

airport_delays.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = airport_delays.head(10)

plt.figure(figsize=(10,5))
plt.bar(top10["origin"], top10["avg_delay"])

plt.title("Top 10 Airports by Average Departure Delay")
plt.xlabel("Airport")
plt.ylabel("Average Delay (minutes)")

plt.tight_layout()
plt.savefig("../figures/airport_delays.png")
plt.show()
```

```python
query = """
SELECT
    route_type,
    ROUND(AVG(dep_delay),2) AS average_delay,
    COUNT(*) AS flights
FROM Flights
GROUP BY route_type
"""

route_summary = pd.read_sql(query, conn)

route_summary
```

```python
query = """
SELECT
    dest,
    ROUND(AVG(arr_delay),2) AS avg_arrival_delay,
    COUNT(*) AS total_flights
FROM Flights
GROUP BY dest
HAVING total_flights > 500
ORDER BY avg_arrival_delay DESC
LIMIT 10
"""

destination_delays = pd.read_sql(query, conn)
destination_delays
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(11,5))

plt.bar(
    destination_delays["dest"],
    destination_delays["avg_arrival_delay"]
)

plt.title("Top 10 Destinations by Average Arrival Delay")
plt.xlabel("Destination Airport")
plt.ylabel("Average Delay (minutes)")

plt.tight_layout()

plt.savefig("../figures/destination_delays.png", dpi=300)

plt.show()
```

```python
query = """
SELECT
    tailnum,
    COUNT(*) AS flights_operated,
    ROUND(AVG(dep_delay),2) AS average_delay
FROM Flights
WHERE tailnum IS NOT NULL
GROUP BY tailnum
HAVING flights_operated > 100
ORDER BY flights_operated DESC
LIMIT 15
"""

aircraft = pd.read_sql(query, conn)

aircraft.head()
```

```python
from sklearn.model_selection import train_test_split

features = flights[[
    "month",
    "day",
    "distance",
    "air_time"
]]

target = flights["delayed"]

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained successfully ✈️")
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)

print(f"Accuracy: {accuracy:.3f}")

confusion_matrix(y_test, predictions)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Target variable
flights["delayed"] = (flights["dep_delay"] > 15).astype(int)

# Features
X = flights[[
    "month",
    "day",
    "distance",
    "air_time"
]]

y = flights["delayed"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print(X_train.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained successfully ✈️")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

pred = model.predict(X_test)

accuracy = accuracy_score(y_test, pred)

print(f"Accuracy: {accuracy:.2%}")

print(classification_report(y_test, pred))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

plt.figure(figsize=(8,4))

plt.bar(
    importance["Feature"],
    importance["Importance"]
)

plt.title("Feature Importance")
plt.ylabel("Importance Score")

plt.tight_layout()

plt.savefig("../figures/feature_importance.png", dpi=300)

plt.show()
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px

# --------------------------------------------------
# Page Configuration
# --------------------------------------------------
st.set_page_config(
    page_title="Qatar Airways Operations Intelligence",
    page_icon="✈️",
    layout="wide"
)

# --------------------------------------------------
# Load Data
# --------------------------------------------------
@st.cache_data
def load_data():
    return pd.read_csv("dashboard/flights_clean.csv")

flights = load_data()

# --------------------------------------------------
# Sidebar Filters
# --------------------------------------------------
st.sidebar.title("Filters")

airport = st.sidebar.selectbox(
    "Origin Airport",
    ["All"] + sorted(flights["origin"].unique().tolist())
)

month = st.sidebar.selectbox(
    "Month",
    ["All"] + sorted(flights["month"].unique().tolist())
)

filtered = flights.copy()

if airport != "All":
    filtered = filtered[filtered["origin"] == airport]

if month != "All":
    filtered = filtered[filtered["month"] == month]

# --------------------------------------------------
# KPIs
# --------------------------------------------------
total = len(filtered)
otp = round((filtered["dep_delay"] <= 15).mean() * 100, 1)
avg_dep = round(filtered["dep_delay"].mean(), 1)
avg_arr = round(filtered["arr_delay"].mean(), 1)

st.title("✈️ Qatar Airways Operations Intelligence")

c1, c2, c3, c4 = st.columns(4)

c1.metric("Flights", f"{total:,}")
c2.metric("On-Time %", f"{otp}%")
c3.metric("Departure Delay", f"{avg_dep} min")
c4.metric("Arrival Delay", f"{avg_arr} min")

st.divider()

# --------------------------------------------------
# Airport Performance
# --------------------------------------------------
airport_summary = (
    filtered
    .groupby("origin", as_index=False)
    .agg(
        average_delay=("dep_delay", "mean"),
        flights=("origin", "count")
    )
)

fig = px.bar(
    airport_summary,
    x="origin",
    y="average_delay",
    color="average_delay",
    hover_data=["flights"],
    title="Average Departure Delay by Airport",
    labels={
        "origin": "Airport",
        "average_delay": "Average Delay (minutes)"
    },
    color_continuous_scale="burg"
)

st.plotly_chart(fig, use_container_width=True)

# --------------------------------------------------
# Monthly Trend
# --------------------------------------------------
monthly = (
    filtered
    .groupby("month", as_index=False)
    .agg(
        average_delay=("dep_delay", "mean")
    )
)

line = px.line(
    monthly,
    x="month",
    y="average_delay",
    markers=True,
    title="Monthly Average Departure Delay"
)

st.plotly_chart(line, use_container_width=True)

# --------------------------------------------------
# Delay Distribution
# --------------------------------------------------
hist = px.histogram(
    filtered,
    x="dep_delay",
    nbins=60,
    title="Distribution of Departure Delays"
)

st.plotly_chart(hist, use_container_width=True)

st.caption("Built using Python, SQL, Plotly and Streamlit")
```
